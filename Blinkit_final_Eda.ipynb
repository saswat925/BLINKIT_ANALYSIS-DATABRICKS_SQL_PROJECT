```python
# Monthly Orders

monthly_orders = (

    df.groupby(
        df["order_date"].dt.to_period("M")
    )["order_id"]
    .count()
    .sort_index()

)

monthly_orders
```

```python
##monthly revenue trend

fig, ax = plt.subplots(figsize=(15,7))

x = range(len(monthly_revenue))
y = monthly_revenue.values

ax.plot(
    x,
    y,
    marker="o"
)

ax.set_xticks(x)
ax.set_xticklabels(
    monthly_revenue.index,
    rotation=45
)

# Labels
for i, value in enumerate(y):

    ax.text(
        i,
        value,
        f"₹{value/100000:.1f}L",
        ha="center",
        va="bottom"
    )

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.grid(alpha=.3)

plt.tight_layout()
plt.show()
```

```python
# Monthly Estimated Profit Trend

fig, ax = plt.subplots(figsize=(18,7))

x = range(len(monthly_profit))
y = monthly_profit.values

ax.plot(
    x,
    y,
    marker="o"
)

ax.set_xticks(x)
ax.set_xticklabels(
    monthly_profit.index,
    rotation=45
)

# Labels
for i, value in enumerate(y):

    ax.text(
        i,
        value,
        f"₹{value/100000:.2f} L",
        ha="center",
        va="bottom"
    )

plt.title("Monthly Estimated Profit Trend")
plt.xlabel("Month")
plt.ylabel("Estimated Profit (₹)")
plt.grid(alpha=.3)

# Labels ke liye extra space
plt.ylim(0, monthly_profit.max()*1.15)

plt.tight_layout()
plt.show()
```
