                           One Table by Merge  for Better EDA 

```python
# Step-1 : Customers + Orders

cust_orders = pd.merge(
    customers,
    orders,
    on="customer_id",
    how="inner"
)

cust_orders.shape
```

```python
# Step-2 : Add Order Items

cust_orders_items = pd.merge(
    cust_orders,
    order_items,
    on="order_id",
    how="inner"
)

cust_orders_items.shape
```

```python
# Step-3 : Add Products

final_df = pd.merge(
    cust_orders_items,
    products,
    on="product_id",
    how="inner"
)

final_df.shape
```

```python
# Revenue
final_df["revenue"] = (
    final_df["quantity"]
    * final_df["unit_price"]
).round(2)
```

```python
# Estimated Cost Price
final_df["cost_price"] = (
    final_df["price"]
    *
    (1-final_df["margin_percentage"]/100)
).round(2)
```

```python
final_df.head(1)
```

```python
# Estimated Profit
final_df["profit"] = (
    final_df["quantity"]
    *
    (final_df["price"]- final_df["cost_price"]).round(2))
```

```python
# Delivery Minutes

final_df["delivery_minutes"] = (

    final_df["actual_delivery_time"]
    -
    final_df["promised_delivery_time"]

).dt.total_seconds()/60
```

```python
final_df[[
    "delivery_status",
    "delivery_minutes"
]].head()
```

```python
final_df.head(1)
```

Business Assumption: The dataset does not provide actual cost price information. Therefore, cost price and profit have been estimated using the provided margin percentage and selling price. These values represent estimated business metrics and should not be interpreted as actual accounting profit.

Inactive Customers extract

```python
inactive_customers = customers[
    ~customers["customer_id"].isin(
        orders["customer_id"]
    )
]

inactive_customers.head()
```

```python
inactive_customers["customer_id"].nunique()
```

```python
##Active Customers
active_customers = customers[
    customers["customer_id"].isin(
        orders["customer_id"]
    )
]

active_customers["customer_id"].nunique()
```

```python
inactive_percentage = round(

    (inactive_customers["customer_id"].nunique()
    /
    customers["customer_id"].nunique())

    *100,2

)

inactive_percentage
```

The Customers table contains 2,500 registered customers, whereas only 2,172 unique customers are present in the Orders table. This indicates that 328 customers did not place any orders during the analysis period. Since the final analytical dataset was created using order-related tables, it includes only active customers who generated transactions.
